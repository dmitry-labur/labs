```python

df = pd.read_csv(
    'billboard_hot_100_spotify_audio_features_enriched.csv'
)

df
```

```python

print(df)
```

```python

top_10 = df.nlargest(10, 'spotify_track_popularity')

result = top_10[['song', 'performer', 'spotify_track_popularity']]

print("ТОП-10 ПОПУЛЯРНЫХ ПЕСЕН:")
print(result.to_string(index=False))
```

```python

for index, row in df.iterrows():
    print("Песня:", row['song'])
    print("Исполнитель:", row['performer'])
    print("Популярность:", row['spotify_track_popularity'])
    print("-------------------------")

    if index >= 9:
        break
```

```python

plt.figure(figsize=(14, 5))

sns.histplot(
    data=df,
    x='spotify_track_popularity',
    bins=30,
    kde=True
)

plt.title('Распределение популярности')
plt.xlabel('Популярность')
plt.ylabel('Количество песен')

plt.show()
```

```python

print(df.isnull().sum())
```

```python

# Заполнение пропусков в числовых столбцах

numeric_columns = df.select_dtypes(include='number').columns
df[numeric_columns] = df[numeric_columns].fillna(0)

# Заполнение пропусков в текстовых столбцах

text_columns = df.select_dtypes(include='object').columns
df[text_columns] = df[text_columns].fillna('Не указано')

# Проверка

print(df.isnull().sum())
```

```python

df.to_csv(
    'billboard_dataset_clean.csv',
    index=False,
    encoding='utf-8-sig'
)

print("Файл успешно сохранён!")
```
